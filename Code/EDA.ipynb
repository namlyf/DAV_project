```python
years  = [2021, 2022, 2023]
colors = ['#4C9BE8', '#F5A623','#E74C3C']

fig, axes = plt.subplots(3, 1, figsize=(16, 8), sharey=True)

for ax, year, color in zip(axes, years, colors):
    daily_sum = df[df['year'] == year].groupby(['month', 'day'])['ALLSKY_SFC_SW_DWN'].sum().reset_index()
    daily_sum['day_of_year'] = range(1, len(daily_sum) + 1)

    ax.plot(daily_sum['day_of_year'], daily_sum['ALLSKY_SFC_SW_DWN'],
            color=color, linewidth=1.2)

    # Đường kẻ dọc phân cách từng tháng
    month_starts = daily_sum.groupby('month')['day_of_year'].first()
    for month, pos in month_starts.items():
        ax.axvline(x=pos, color='gray', linestyle='--', linewidth=0.7, alpha=0.5)
        ax.text(pos + 3, ax.get_ylim()[1] * 0.92 if ax.get_ylim()[1] > 0 else 5000,
                f'T{month}', fontsize=9, color='gray')

    ax.set_title(f'Tổng GHI theo ngày — năm {year} — Hà Nội', fontsize=13, fontweight='bold')
    ax.set_xlabel('Ngày trong năm', fontsize=11)
    ax.set_ylabel('Tổng GHI (Wh/m²)', fontsize=11)
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Các cột cần tính độ lệch chuẩn
cols = ["T2M", "RH2M", "PRECTOTCORR", "WS10M"]

# Tính std
std_values = df[cols].std()

# Vẽ biểu đồ
plt.figure(figsize=(8, 5), dpi=120)

ax = sns.barplot(
    x=std_values.index,
    y=std_values.values
)

# Hiển thị giá trị trên đầu cột
for i, v in enumerate(std_values.values):
    ax.text(
        i,
        v,
        f"{v:.2f}",
        ha="center",
        va="bottom",
        fontsize=11
    )

plt.title("Standard Deviation of Weather Features", fontsize=14)
plt.xlabel("Feature")
plt.ylabel("Standard Deviation")
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
df_hanoi = pd.read_csv(r'E:\2025.2\DAV\data\Power_solar_Hanoi.csv')
df_bacgiang = pd.read_csv(r'E:\2025.2\DAV\data\Power_solar_BacGiang.csv')
df_sonla = pd.read_csv(r'E:\2025.2\DAV\data\Power_solar_SonLa.csv')

```

```python
def filter_daytime(df, night_filter=True, sza_threshold=90, verbose=True):
    """Return daytime rows filtered by solar zenith angle (SZA)."""
    if night_filter:
        df_day = df[df['SZA'] < sza_threshold].copy()
        df_day.reset_index(drop=True, inplace=True)
        if verbose:
            print(f"\n[NIGHT_FILTER = True]")
            print(f"Trước lọc : {len(df):,} rows")
            print(f"Sau lọc   : {len(df_day):,} rows")
            print(f"Đã loại   : {len(df) - len(df_day):,} rows ban đêm "
                  f"({(len(df) - len(df_day)) / len(df) * 100:.1f}%)")
    else:
        df_day = df.copy()
        if verbose:
            print(f"\n[NIGHT_FILTER = False] Giữ toàn bộ {len(df_day):,} rows")

    if verbose:
        print(f"\nShape sau xử lý: {df_day.shape}")
    return df_day

# Example usage
NIGHT_FILTER = True
SZA_THRESHOLD = 90

df_hanoi_day = filter_daytime(df_hanoi, night_filter=NIGHT_FILTER, sza_threshold=SZA_THRESHOLD)
df_bacgiang_day = filter_daytime(df_bacgiang, night_filter=NIGHT_FILTER, sza_threshold=SZA_THRESHOLD)
df_sonla_day = filter_daytime(df_sonla, night_filter=NIGHT_FILTER, sza_threshold=SZA_THRESHOLD)
```

```python
# Coefficient of Variation (CV) của 4 feature theo từng khu vực
features = ['T2M', 'RH2M', 'PRECTOTCORR', 'WS10M']
regions = [
    ('Hà Nội', df_hanoi_day),
    ('Bắc Giang', df_bacgiang_day),
    ('Sơn La', df_sonla_day)
]

cv_rows = []
for region_name, region_df in regions:
    stats = region_df[features].agg(['mean', 'std']).T
    stats['cv'] = (stats['std'] / stats['mean']) * 100
    for feature, row in stats.iterrows():
        cv_rows.append({
            'region': region_name,
            'feature': feature,
            'cv': row['cv']
        })

cv_df = pd.DataFrame(cv_rows)

plt.figure(figsize=(10, 6), dpi=120)
sns.barplot(data=cv_df, x='feature', y='cv', hue='region', palette='Set2')
plt.title('Coefficient of Variation (%) của 4 feature theo khu vực', fontsize=14, fontweight='bold')
plt.xlabel('Feature', fontsize=12)
plt.ylabel('CV (%)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.legend(title='Khu vực')
plt.tight_layout()
plt.show()
```

```python
# So sánh 3 khu vực qua 4 feature: T2M, RH2M, PRECTOTCORR, WS10M
regions = [
    ('Hà Nội', df_hanoi_day),
    ('Bắc Giang', df_bacgiang_day),
    ('Sơn La', df_sonla_day)
]
variables = ['T2M', 'RH2M', 'PRECTOTCORR', 'WS10M']

combined = pd.concat(
    [region_df[variables].assign(region=region_name) for region_name, region_df in regions],
    ignore_index=True
)

fig, axes = plt.subplots(nrows=len(variables), ncols=1, figsize=(12, 18), sharex=False)
for ax, var in zip(axes, variables):
    sns.boxplot(
        x=var,
        y='region',
        data=combined,
        ax=ax,
        palette='Set2',
        orient='h'
    )
    ax.set_title(f'{var} theo khu vực', fontsize=13, fontweight='bold')
    ax.set_xlabel('Giá trị', fontsize=11)
    ax.set_ylabel('Khu vực', fontsize=11)
    ax.grid(axis='x', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
